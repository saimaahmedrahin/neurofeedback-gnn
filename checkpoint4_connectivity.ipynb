```python
from pathlib import Path
import numpy as np
import nibabel as nib

# Full path to data
data_path = Path("C:/Users/rahin/OneDrive/Desktop/TU/Summer 2026/Brain fmri in R/Selim code/Processed Rest1&2 Scans/Processed rest scans/E3746")
brik_file = list(data_path.glob("*.BRIK"))[0]

print(f"Loading {brik_file.name}...")
img = nib.load(brik_file)
fmri_data = img.get_fdata()

# Simple approach: just get mean activity per voxel and compute on that
# (This avoids the huge memory issue)
print("Computing connectivity on voxel clusters...")

# Reshape to 2D: (timepoints, voxels)
voxels = fmri_data.reshape(-1, fmri_data.shape[-1]).T

print(f"Shape: {voxels.shape} (timepoints={voxels.shape[0]}, voxels={voxels.shape[1]})")

# Sample 1000 random voxels to keep memory low
np.random.seed(42)
sample_idx = np.random.choice(voxels.shape[1], 1000, replace=False)
ts_sample = voxels[:, sample_idx]

print(f"Using {ts_sample.shape[1]} sample voxels...")

# Compute correlation
corr_matrix = np.corrcoef(ts_sample.T)
adj_matrix = (np.abs(corr_matrix) > 0.6).astype(int)
np.fill_diagonal(adj_matrix, 0)

degree = np.sum(adj_matrix, axis=0)
n_edges = np.sum(adj_matrix) // 2
density = n_edges / (len(degree) * (len(degree)-1) / 2)

print(f"\n=== Subject E3746 (1000 voxel sample) ===")
print(f"Edges (r > 0.6): {n_edges}")
print(f"Mean degree: {np.mean(degree):.2f}")
print(f"Network density: {density:.4f}")
```

```python
from pathlib import Path
import numpy as np
import nibabel as nib

# Use FULL path to data
data_path = Path("C:/Users/rahin/OneDrive/Desktop/TU/Summer 2026/Brain fmri in R/Selim code/Processed Rest1&2 Scans/Processed rest scans/E3746")

print(f"Looking for: {data_path}")
print(f"Exists? {data_path.exists()}")

if data_path.exists():
    brik_file = list(data_path.glob("*.BRIK"))[0]
    print(f"Found: {brik_file.name}")
    
    img = nib.load(brik_file)
    fmri_data = img.get_fdata()
    
    mask = np.var(fmri_data, axis=3) > 0
    brain_voxels = np.where(mask)
    ts = fmri_data[brain_voxels].T
    
    corr_matrix = np.corrcoef(ts.T)
    adj_matrix = (np.abs(corr_matrix) > 0.6).astype(int)
    np.fill_diagonal(adj_matrix, 0)
    
    degree = np.sum(adj_matrix, axis=0)
    n_edges = np.sum(adj_matrix) // 2
    density = n_edges / (len(degree) * (len(degree)-1) / 2)
    
    print(f"\n=== Subject E3746 ===")
    print(f"Brain voxels: {len(brain_voxels[0])}")
    print(f"Mean correlation: {np.mean(corr_matrix[np.triu_indices_from(corr_matrix, k=1)]):.3f}")
    print(f"Edges (r > 0.6): {n_edges}")
    print(f"Mean degree: {np.mean(degree):.2f}")
    print(f"Network density: {density:.4f}")
else:
    print("PATH NOT FOUND - check the folder path")
```
