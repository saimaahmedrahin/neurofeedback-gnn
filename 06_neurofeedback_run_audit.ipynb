# Neurofeedback Run Audit

## Goal

Determine which neurofeedback runs are available for the final imaging cohort
before performing any new analysis.

Potential runs:
- baseline
- practice
- training 1
- training 2
- training 3
- transfer

No modeling is performed in this notebook at this stage.

```python
import pandas as pd
from pathlib import Path

PARTICIPANTS_FILE = Path(
    r"C:\Neurofeedback_Data_Fall26\Imaging\Processed Rest1&2 Scans\participants.tsv"
)

df = pd.read_csv(PARTICIPANTS_FILE, sep="\t")

run_cols = [
    "Sub",
    "Exam",
    "Group",
    "rtbaseline",
    "rtpractice",
    "rttraining1",
    "rttraining2",
    "rttraining3",
    "rttransfer",
]

print("Dataset shape:", df.shape)
print("\nRelevant columns:")
print(df[run_cols].head(20).to_string(index=False))
```

```python
FINAL_EXAMS = [
    "E3746", "E3799", "E3973", "E4051", "E4209", "E4253",
    "E4324", "E4350", "E4360", "E4484", "E4689", "E4697",
    "E4745", "E5215", "E5580", "E5586", "E5693", "E5694"
]

nf_cols = [
    "rtpractice",
    "rttraining1",
    "rttraining2",
    "rttraining3",
    "rttransfer"
]

final_nf = df[df["Exam"].astype(str).isin(FINAL_EXAMS)][
    ["Sub", "Exam", "Group"] + nf_cols
].copy()

print("Final participants found:", len(final_nf))
print("\nRun availability:")
print(final_nf[nf_cols].notna().sum())

print("\nParticipant-level table:")
print(final_nf.to_string(index=False))
```

```python
from pathlib import Path

ROOT = Path(r"C:\Neurofeedback_Data_Fall26")

keywords = [
    "training",
    "practice",
    "transfer",
    "scan_6",
    "scan_7",
    "E3746"
]

for keyword in keywords:
    matches = [
        p for p in ROOT.rglob("*")
        if keyword.lower() in p.name.lower()
    ]

    print(f"\n--- {keyword}: {len(matches)} matches ---")
    for p in matches[:20]:
        print(p)
```
